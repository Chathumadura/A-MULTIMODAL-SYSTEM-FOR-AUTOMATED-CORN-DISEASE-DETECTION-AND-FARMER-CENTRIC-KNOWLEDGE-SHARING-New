# Preprocess Sri Lanka Corn Dataset to Anuradhapura 1500 Records

This notebook loads the 10,000-record Sri Lanka corn dataset, filters only Anuradhapura, removes additional columns, optionally removes season, and exports a 1,500-record dataset for model training.

Reference links used to justify source-informed construction:
- Department of Census and Statistics agriculture statistics: https://www.statistics.gov.lk/Agriculture/StaticalInformation
- Cereal Production Statistics dataset page: https://data.gov.lk/dataset/cereal-production-statistics-maha-season-2011/resource/8b5acd03-8196-4961-9f57-4d0ff21edd1f
- Department of Agriculture AgStat: https://doa.gov.lk/wp-content/uploads/2020/05/AgstatBK.pdf
- Crop Forecast report: https://doa.gov.lk/wp-content/uploads/2020/02/crop-forcast.pdf
- Maize Crop Yield Estimation Survey 2023/24 Maha: https://sapp.lk/wp-content/uploads/2025/10/Maize-CCS-Report.pdf


```python
import pandas as pd
from pathlib import Path

# Define input and output paths
data_path = Path("output/sri_lanka_corn_dataset_10000.csv")
out_dir = Path("output")
out_dir.mkdir(exist_ok=True)

# Load the primary Sri Lanka wide dataset (10,000 records)
df = pd.read_csv(data_path)
df.head()
```

```python
# Filter for Anuradhapura district only to focus the model on regional agricultural patterns
anur = df[df["district"] == "Anuradhapura"].copy()
print(f"Anuradhapura records: {anur.shape[0]}")
```

```python
# Remove auxiliary columns that are not part of the final model's predictor set
columns_to_remove = [
    "province",
    "agro_climatic_zone",
    "season",
    "avg_temperature_c",
    "avg_relative_humidity_pct",
    "planting_density_plants_per_acre",
    "seed_rate_kg_per_acre",
    "soil_ph",
    "soil_organic_matter_pct"
]
anur_model = anur.drop(columns=columns_to_remove)
print(f"Final feature count: {anur_model.shape[1]}")
```

```python
# Downsample to 1,500 records to create the final training dataset artifact
anur_1500 = anur_model.sample(n=1500, random_state=42, replace=False if len(anur_model) >= 1500 else True)

# Save the cleaned regional dataset
anur_1500.to_csv(out_dir / "anuradhapura_corn_dataset_1500.csv", index=False)
anur_1500.head()
```

```python
# Generate a summary of the processed data for validation and reference
anur_1500.describe(include="all").transpose().to_csv(out_dir / "anuradhapura_1500_summary.csv")

print(f"Exported: {out_dir / 'anuradhapura_corn_dataset_1500.csv'}")
print(f"Exported: {out_dir / 'anuradhapura_1500_summary.csv'}")
```
